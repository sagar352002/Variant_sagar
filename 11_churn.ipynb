```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn
%matplotlib inline
```

```python
df=pd.read_csv("Churn_Modelling.csv")
df
```

```python
df["Exited"].value_counts()
df["Gender"].value_counts()
df["Geography"].value_counts()
```

```python
df.drop(columns=["RowNumber","CustomerId","Surname"], inplace=True)
```

```python
df
```

```python
df=pd.get_dummies(df,columns=["Geography","Gender"],drop_first=True)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Split the data into features (X) and target variable (y)
X = df.drop(columns=["Exited"])
y = df["Exited"]

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# Initialize StandardScaler
scaler = StandardScaler()

# Scale the training data
X_train_scaled = scaler.fit_transform(X_train)

# Scale the testing data using the same scaler from training data
X_test_scaled = scaler.transform(X_test)
```

# scale during weight trainign for better convergence

```python
# Start makin model
```

```python
import tensorflow as tf
from tensorflow import keras

# Corrected import for Sequential
from tensorflow.keras.models import Sequential

# Corrected import for Dense
from tensorflow.keras.layers import Dense
```

```python
# Initialize the Sequential model
model = Sequential()

# Add the first Dense layer with 3 units, sigmoid activation, and input dimension 11
model.add(Dense(3, activation="sigmoid", input_dim=11))

# Add the second Dense layer with 1 unit and sigmoid activation
model.add(Dense(1, activation="sigmoid"))
```

```python
model.summary()
```

```python
model.compile(loss="binary_crossentropy",optimizer="adam")
```

```python
model.fit(X_train_scaled,y_train,epochs=10,batch_size=800,validation_split=0.2)
```

```python
model.layers[0].get_weights()
```

```python
model.layers[1].get_weights()
```

```python
y_log=model.predict(X_test)
```

```python
y_pred=np.where(y_log>0.5,1,0)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
